```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel('sample_data/Natural-Disaster.xlsx')
df.head()
```

```python
# Ver las primeras 10 filas
df.head(10)
```

```python
# Ver cuántas filas/columnas tiene
print(df.shape)
```

```python
# Ver los nombres de columnas y tipos de datos
df.info()
```

```python
# Estadísticas básicas (promedios, mínimos, máximos, etc.)
df.describe()
```

```python
# ── 1. ¿Qué tipos de desastres hay? ──────────────────────────────
print(df['Disaster Type'].value_counts().head(10))
```

```python
# ── 2. ¿Cuántos desastres por año? ───────────────────────────────
df.groupby('Start Year')['DisNo.'].count().plot(
    figsize=(14,4), title='Desastres por año', color='steelblue'
)
plt.tight_layout()
plt.show()
```

```python
# ── 3. ¿Qué países tienen más desastres? ─────────────────────────
print(df['Country'].value_counts().head(40))
```

```python
# ── 4. Muertes totales por tipo de desastre ───────────────────────
df.groupby('Disaster Type')['Total Deaths'].sum()\
  .sort_values(ascending=False).head(10)\
  .plot(kind='barh', figsize=(8,5), title='Muertes por tipo de desastre')
plt.tight_layout()
plt.show()
```

```python
# ── 5. Mapa de calor de valores nulos ────────────────────────────
plt.figure(figsize=(14,5))
sns.heatmap(df.isnull(), yticklabels=False, cbar=False, cmap='viridis')
plt.title('Valores nulos por columna')
plt.tight_layout()
plt.show()
```
